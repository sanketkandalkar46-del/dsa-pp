Knowledge Representation and Expert System

1. Problem StatementDevelop a rule-based expert system using propositional logic and implement Forward Chaining and Backward Chaining mechanisms for intelligent decision-making (e.g., Medical / Troubleshooting Diagnosis).2. ObjectivesRepresent domain knowledge using IF-THEN production rules.Implement Forward Chaining (data-driven reasoning).Implement Backward Chaining (goal-driven reasoning).3. Key ConceptsMechanismReasoning TypeOperationForward ChainingData-drivenStarts with known facts and applies IF-THEN rules iteratively to derive new facts/conclusions.Backward ChainingGoal-drivenStarts with a hypothesis/goal and works backward to verify if known facts support it.4. Knowledge Base (Rules)Rule 1: IF fever AND cough THEN flu_symptomsRule 2: IF flu_symptoms AND fatigue THEN influenzaRule 3: IF rash AND fever THEN measles

Box 1: Knowledge Base Class

```python
class KnowledgeBase:
    """Stores facts and IF-THEN rules for the Expert System."""
    def __init__(self):
        self.rules = [
            {'antecedents': {'fever', 'cough'}, 'consequent': 'flu_symptoms'},
            {'antecedents': {'flu_symptoms', 'fatigue'}, 'consequent': 'influenza'},
            {'antecedents': {'rash', 'fever'}, 'consequent': 'measles'}
        ]
```

Box 2: Forward Chaining Engine

```python
class ForwardChainingEngine:
    """Data-driven inference engine."""
    def __init__(self, kb):
        self.kb = kb

    def infer(self, initial_facts):
        facts = set(initial_facts)
        added_new_fact = True

        print("--- Forward Chaining Execution ---")
        print(f"Initial Facts: {facts}")

        while added_new_fact:
            added_new_fact = False
            for rule in self.kb.rules:
                if rule['antecedents'].issubset(facts) and rule['consequent'] not in facts:
                    facts.add(rule['consequent'])
                    print(f"[RULE FIRED] IF {rule['antecedents']} THEN -> {rule['consequent']}")
                    added_new_fact = True

        print(f"Final Derived Facts: {facts}\n")
        return facts
```

Box 3: Backward Chaining Engine

```python
class BackwardChainingEngine:
    """Goal-driven inference engine."""
    def __init__(self, kb):
        self.kb = kb

    def check_goal(self, goal, facts, visited=None):
        if visited is None:
            visited = set()

        if goal in facts:
            return True

        if goal in visited:
            return False

        visited.add(goal)

        for rule in self.kb.rules:
            if rule['consequent'] == goal:
                all_subgoals_proven = True
                for antecedent in rule['antecedents']:
                    if not self.check_goal(antecedent, facts, visited):
                        all_subgoals_proven = False
                        break

                if all_subgoals_proven:
                    return True

        return False
```

Box 4: Execution & Testing

```python
# Initialize Knowledge Base and Engines
kb = KnowledgeBase()
fc_engine = ForwardChainingEngine(kb)
bc_engine = BackwardChainingEngine(kb)

print("==================================================")
print("        EXPERT SYSTEM INFERENCE ENGINE")
print("==================================================\n")

# 1. Test Forward Chaining
known_facts = {'fever', 'cough', 'fatigue'}
derived_facts = fc_engine.infer(known_facts)

# 2. Test Backward Chaining (Proven Goal)
target_goal = 'influenza'
print("--- Backward Chaining Execution ---")
print(f"Checking Hypothesis/Goal: '{target_goal}' with Facts: {known_facts}")
is_proven = bc_engine.check_goal(target_goal, known_facts)
print(f"Goal '{target_goal}' Proven: {is_proven}\n")

# 3. Test Backward Chaining (Unproven Goal)
unproven_goal = 'measles'
print(f"Checking Hypothesis/Goal: '{unproven_goal}' with Facts: {known_facts}")
is_proven_measles = bc_engine.check_goal(unproven_goal, known_facts)
print(f"Goal '{unproven_goal}' Proven: {is_proven_measles}")
print("-" * 50)
```

1. Forward Chaining (Data-Driven)Start: Takes your input symptoms (cough, fatigue, fever).Step 1: Finds matching rule $\rightarrow$ fever + cough = flu_symptoms.Step 2: Uses new fact $\rightarrow$ flu_symptoms + fatigue = influenza.End: Output contains all derived facts including the final diagnosis (influenza)
.2. Backward Chaining (Goal-Driven)
Goal 1 (influenza):Checks required facts $\rightarrow$ Needs flu_symptoms + fatigue.Traces back $\rightarrow$ Needs fever + cough for flu_symptoms.Result: True (All required symptoms exist in input data).
Goal 2 (measles):Checks required facts $\rightarrow$ Needs rash + fever.Traces back $\rightarrow$ Finds fever, but rash is missing.Result: False (Cannot prove measles without rash).
