Code Cell 1 — Import Libraries

```python
import random
import time
```

Code Cell 2 — Define the Environment

```python
class SmartHomeEnvironment:
    def __init__(self):
        self.temperature = 28.0
        self.light_level = 300
        self.presence = True

    def get_sensor_readings(self):
        return {
            'temperature': self.temperature,
            'light_level': self.light_level,
            'presence': self.presence
        }
```

Code Cell 3 — Define the Intelligent Agent

```python
class SmartHomeAgent:
    def __init__(self):
        self.target_temp = 22
        self.min_light = 500

    def perceive_and_act(self, percepts):
        temp = percepts['temperature']
        light = percepts['light_level']
        presence = percepts['presence']

        if not presence:
            return "AC OFF", "LIGHT OFF"

        if temp > self.target_temp + 1:
            ac = "AC COOLING"
        elif temp < self.target_temp - 1:
            ac = "AC HEATING"
        else:
            ac = "AC OFF"

        if light < self.min_light:
            light_action = "LIGHT ON"
        else:
            light_action = "LIGHT OFF"

        return ac, light_action
```

Code Cell 4 — Test the Intelligent Agent

```python
env = SmartHomeEnvironment()
agent = SmartHomeAgent()

sensor_data = env.get_sensor_readings()
print("Sensor Readings:", sensor_data)

ac_action, light_action = agent.perceive_and_act(sensor_data)
print("AC Action:", ac_action)
print("Light Action:", light_action)
```
