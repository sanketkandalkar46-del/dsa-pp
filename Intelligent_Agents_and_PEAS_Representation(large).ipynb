Code Block 1: Imports & Environment Clas

```python
import time
import random

class SmartHomeEnvironment:
    """Simulates the physical room state and sensor hardware."""
    def __init__(self, initial_temp=28.0, initial_light=300, initial_presence=True):
        self.temperature = initial_temp
        self.light_level = initial_light
        self.presence = initial_presence

    def get_sensor_readings(self):
        """Returns current sensor percepts."""
        return {
            'temperature': self.temperature,
            'light_level': self.light_level,
            'presence': self.presence
        }

    def update_state(self, ac_action, light_action):
        """Simulates environment feedback loop based on actuator execution."""
        if ac_action == "AC COOLING":
            self.temperature -= 1.5
        elif ac_action == "AC HEATING":
            self.temperature += 1.5

        if light_action == "LIGHT ON":
            self.light_level = 800
        else:
            self.light_level = 200
```

Code Block 2: Agent Program Class

```python
class SmartHomeAgent:
    """Implements rule-based decision-making logic."""
    def __init__(self, target_temp=22.0, min_light=500):
        self.target_temp = target_temp
        self.min_light = min_light

    def perceive_and_act(self, percepts):
        """Maps sensor percepts to actuator actions."""
        temp = percepts['temperature']
        light = percepts['light_level']
        presence = percepts['presence']

        # Rule 1: Energy Saving Mode when no presence
        if not presence:
            return "AC OFF", "LIGHT OFF"

        # Rule 2: Temperature Control Logic
        if temp > self.target_temp + 1.0:
            ac_action = "AC COOLING"
        elif temp < self.target_temp - 1.0:
            ac_action = "AC HEATING"
        else:
            ac_action = "AC OFF"

        # Rule 3: Lighting Control Logic
        if light < self.min_light:
            light_action = "LIGHT ON"
        else:
            light_action = "LIGHT OFF"

        return ac_action, light_action
```

Code Block 3: Simulation Execution Loop

```python
# Initialize Environment and Agent
env = SmartHomeEnvironment(initial_temp=28.0, initial_light=300, initial_presence=True)
agent = SmartHomeAgent(target_temp=22.0, min_light=500)

print("==================================================")
print("     SMART HOME INTELLIGENT AGENT EXECUTION")
print("==================================================\n")

# Run Multi-Step Simulation Loop
for step in range(1, 5):
    print(f"--- Iteration {step} ---")

    # 1. Sense Environment
    percepts = env.get_sensor_readings()
    print(f"[SENSORS]   Temp: {percepts['temperature']:.1f}°C | Light: {percepts['light_level']} Lux | Presence: {percepts['presence']}")

    # 2. Decide Actions
    ac_action, light_action = agent.perceive_and_act(percepts)
    print(f"[ACTUATORS] Decision -> AC: {ac_action} | Lighting: {light_action}")

    # 3. Apply Actions to Environment
    env.update_state(ac_action, light_action)
    print("-" * 50)
```
