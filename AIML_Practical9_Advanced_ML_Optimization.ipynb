**Aim:** To implement and compare SVM, Random Forest and AdaBoost algorithms. Apply 5-fold cross-validation and Grid Search for model optimization.

**Dataset:** Breast Cancer Wisconsin dataset (available in Scikit-learn). No CSV upload required.

## 1. Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

## 2. Load dataset
Expected: 569 records and 30 input features.

```python
data = load_breast_cancer()
x = data.data
y = data.target
print(x.shape)
print(y.shape)
```

## 3. Split dataset
80% training, 20% testing.

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

## 4. Create models

```python
svm = Pipeline([
    ('scaler', StandardScaler()),
    ('model', SVC())
])

rf = RandomForestClassifier(random_state=42)
ada = AdaBoostClassifier(random_state=42)
```

## 5. Train and compare models (before optimization)

```python
models = [svm, rf, ada]
names = ['SVM', 'Random Forest', 'AdaBoost']

for model, name in zip(models, names):
    model.fit(x_train, y_train)
    pred = model.predict(x_test)
    print(name)
    print("Accuracy:", accuracy_score(y_test, pred))
    print("Precision:", precision_score(y_test, pred))
    print("Recall:", recall_score(y_test, pred))
    print("F1 Score:", f1_score(y_test, pred))
    print()
```

## 6. 5-fold cross-validation
The dataset is divided into 5 parts. The model trains and tests 5 times using different parts.

```python
for model, name in zip(models, names):
    score = cross_val_score(model, x, y, cv=5)
    print(name, ":", score.mean())
```

## 7. Grid Search – SVM

```python
params = {
    'model__C': [0.1, 1, 10],
    'model__kernel': ['linear', 'rbf']
}

grid = GridSearchCV(svm, params, cv=5)
grid.fit(x_train, y_train)
print(grid.best_params_)
```

## 8. Grid Search – Random Forest

```python
params = {
    'n_estimators': [50, 100],
    'max_depth': [None, 5, 10]
}

grid_rf = GridSearchCV(rf, params, cv=5)
grid_rf.fit(x_train, y_train)
print(grid_rf.best_params_)
```

## 9. Grid Search – AdaBoost

```python
params = {
    'n_estimators': [50, 100],
    'learning_rate': [0.1, 1]
}

grid_ada = GridSearchCV(ada, params, cv=5)
grid_ada.fit(x_train, y_train)
print(grid_ada.best_params_)
```

## 10. Compare optimized models

```python
best_models = [
    grid.best_estimator_,
    grid_rf.best_estimator_,
    grid_ada.best_estimator_
]

for model, name in zip(best_models, names):
    pred = model.predict(x_test)
    print(name)
    print("Accuracy:", accuracy_score(y_test, pred))
    print("F1 Score:", f1_score(y_test, pred))
    print()
```

## 11. Plot accuracy comparison

```python
acc = []
for model in best_models:
    pred = model.predict(x_test)
    acc.append(accuracy_score(y_test, pred))

plt.bar(names, acc)
plt.xlabel("Algorithms")
plt.ylabel("Accuracy")
plt.title("Model Comparison")
plt.show()
```

## Viva – Important Terms
| Term | Simple meaning |
|---|---|
| SVM | Finds the best boundary between classes. |
| Random Forest | Combines multiple decision trees. |
| AdaBoost | Combines weak learners to improve prediction. |
| Cross-validation | Tests a model using different data parts. |
| Grid Search | Finds the best parameter combination. |
| Accuracy | Correct predictions out of total predictions. |
| Precision | How many predicted positives are correct. |
| Recall | How many actual positives are found. |
| F1 Score | Balance between precision and recall. |
